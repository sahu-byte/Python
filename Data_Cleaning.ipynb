```python
from urllib.parse import quote_plus
passcode = quote_plus('sahu20031004@gmail.com')

from sqlalchemy import create_engine
engine = create_engine(f"mysql+pymysql://root:{passcode}@localhost/readme_uncleaned")

print(engine.url)

query = 'select * from raw_employee_data'
employee_data = pd.read_sql(query,engine)
employee_data
```

```python
employee_data.info()
employee_data.isna().sum()
```

```python
print(employee_data.describe(include = 'object'))
print(employee_data.describe())
```

```python
employee_data['Age'] = employee_data['Age'].fillna(employee_data['Age'].median())
```

```python
employee_data['Age'] = employee_data['Age'].mask(employee_data['Age'] > 99, employee_data['Age'].median())
employee_data['Age'] = employee_data['Age'].mask(employee_data['Age'] < 18, employee_data['Age'].median())
```

```python
employee_data['Gender'] = employee_data['Gender'].fillna(employee_data['Gender'].mode()[0])
```

```python
employee_data['Gender'] = employee_data['Gender'].str.strip()
employee_data['Gender'] = employee_data['Gender'].str.lower()
employee_data['Gender'] = employee_data['Gender'].str.title()
employee_data['Gender'] = employee_data['Gender'].replace({'' : 'Male'})
```

```python
employee_data['Gender'].unique()
```

```python
employee_data[employee_data['Department'].isna()]
employee_data['Department'].unique()
employee_data['Department'] = employee_data['Department'].str.strip()
employee_data['Department'] = employee_data['Department'].str.upper()
employee_data['Department'] = employee_data['Department'].str.lower()
employee_data['Department'] = employee_data['Department'].str.title()
employee_data['Department'] = employee_data['Department'].replace({'Cs':'Customer Support','It':'Information Technology','I.T.':'Information Technology','Cust. Support':'Customer Support','Mktg':'Marketing','Sale':'Sales','Hr':'Human Resources','Fin':'Finance','H.R.':'Human Resources'})
```

```python
employee_data['Department'].value_counts()
employee_data['Department'] = employee_data['Department'].fillna(employee_data['Department'].mode()[0])
```

```python
employee_data['Department'].value_counts()
employee_data['Department'] = employee_data['Department'].replace({'Missing':'Customer Support','Na':'Marketing','-':'Information Technology',' ':'Customer Support','':'Customer Support'})
```

```python
employee_data['Department'].info()
```

```python
employee_data['City'].value_counts()
employee_data['City'] = employee_data['City'].str.strip()
employee_data['City'] = employee_data['City'].str.upper()
employee_data['City'] = employee_data['City'].str.lower()
employee_data['City'] = employee_data['City'].str.title()
```

```python
employee_data['City'].value_counts()
employee_data['City'] = employee_data['City'].replace({'Bengaluru':'Bangalore','Calcutta':'Kolkata','Bombay':'Mumbai','':'Bangalore','?':'Pune','Na':'Hyderabad','Missing':'Kolkata','Mumai':'Mumbai','N/A':'Delhi'})
```

```python
employee_data['City'] = employee_data['City'].fillna(employee_data['City'].mode()[0])
employee_data['City']
```

```python
employee_data['Salary'] = employee_data['Salary'].fillna(employee_data['Salary'].mean())
employee_data['Salary'].isna().sum()
```

```python
employee_data['Joining_Date'].value_counts()
employee_data['Joining_Date'] = employee_data['Joining_Date'].str.strip()
```

```python
employee_data['Joining_Date'] = employee_data['Joining_Date'].str.replace('.', '-')
employee_data['Joining_Date'] = employee_data['Joining_Date'].str.replace('/', '-')
```

```python
employee_data['Joining_Date'].value_counts()
employee_data['Joining_Date'] = pd.to_datetime(
    employee_data['Joining_Date'],
    format='mixed',
    dayfirst=True,
    errors='coerce')
```

```python
employee_data['Joining_Date'] = employee_data['Joining_Date'].fillna(employee_data['Joining_Date'].bfill())
```

```python
employee_data['Performance_Rating'] = employee_data['Performance_Rating'].fillna(employee_data['Performance_Rating'].mean())
```

```python
employee_data['Email'] = employee_data['Email'].fillna('Not Available')
```

```python
employee_data['Email'] = employee_data['Email'].replace({'NA':'Not Available','-':'Not Available'})
```

```python
employee_data['Email'].value_counts()
```
